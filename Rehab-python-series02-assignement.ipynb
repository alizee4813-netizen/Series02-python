# 1. <a id='toc1_'></a>[Create and improve a notebook to plot sinusoidal signals](#toc0_)

You have created a Jupyter Notebook to plot sinusoidal signals, you used branches to improve your notebook, and you are now satisfied with the way you explained the theory and and coded the implementation. 

What you now need is to share this work with a teammate for review. After the review, you will use the feedback to improve your notebook and, when you and the reviewer are satisfied with the result, you will make your work public.

To share your work with a teammate for review, Github has a nice feature to do this, called "pull request". 

You will need **to work in pairs**, one being the *author* of the notebook and the other being the *reviewer* (et vice versa).

The working logic is as follows: 
- The author creates a pull request to merge the final branch into the main branch and assigns the reviewer to this pull request.
- The reviewer reads the notebook and adds comments to the pull request.
- The author uses the comments to improve the notebook and pushes the changes to the final branch.
- When both the author and the reviewer are satisfied with the result, the reviewer approves the pull request and the author merges the final branch into the main branch.

How to do it? Here are some tips:
1. Add your colleague as a collaborator to your GitHub repository (if you have not done it yet).
1. Create a pull request (PR) for your changes, including a description of what you have done and what you would like feedback on.
3. In the PR page on GitHub, click on "Request a review" from your colleague.
4. Be open to feedback and ready to make changes based on the review.

# 2. <a id='toc2_'></a>[Reproducibility of research and quality of code](#toc0_)

Reproducibility of research is the ability to reproduce the results of a study using the same data and methods as the original study. It is a fundamental principle of scientific research, which depends on the quality of the code used to analyze the data.
Hence, the quality of researcher's code is a very important topic in data analysis for scientific research (as it is in software engineering).  

This is very well explained in [The Turing Way](https://book.the-turing-way.org/). Take a look at the chapter about **Code Quality** (in the section about reproducible research). 


Paraphrased from https://www.codegrip.tech/productivity/what-is-code-quality-why-is-it-important/  

> Code quality measures of how well can code communicate between *researchers*. Similar analyses may have an evident difference in code from *researchers* to *researchers*, but what matters most is that the quality of code remains the same throughout. A quality code is one that is clear, simple, well tested, bug-free, refactored, documented, and performant. But the primary measure of high-quality code in compliance with the specification depends on the needs of the *research*. 

**Reproducibility of research** and **quality of code** are the two criterions for the reviewer to assess the notebook and suggest improvements.

```python
# https://nbconvert.readthedocs.io/en/latest/removing_cells.html

# notebook_file_name = "Untitled-1"

# https://github.com/msm1089/ipynbname/issues/17#issuecomment-1293269863
# from IPython import get_ipython

# ip = get_ipython()
# path = None
# if "__vsc_ipynb_file__" in ip.user_ns:
#     path = ip.user_ns["__vsc_ipynb_file__"]


# notebook_file_name = path.split("/")[-1].split(".")[0]

from traitlets.config import Config
import nbformat as nbf
from nbconvert.exporters import HTMLExporter
from nbconvert.preprocessors import TagRemovePreprocessor
from IPython import get_ipython


def get_notebook_name():
    """
    Get the current notebook name (without extension).
    """
    ip = get_ipython()
    path = None
    if "__vsc_ipynb_file__" in ip.user_ns:
        path = ip.user_ns["__vsc_ipynb_file__"]

    return path.split("/")[-1].split(".")[0]

# path
# output/toto.ipynb
# path.split("/") -> ['output', 'toto.ipynb']
# [-1]  -> 'toto.ipynb'
# split(".") -> 'toto', 'ipynb'
# [0] -> 'toto'
pop^
#

# Get the notebook name
notebook_file_name = get_notebook_name()


# Setup config
c = Config()

# Configure tag removal - be sure to tag your cells to remove  using the
# words remove_cell to remove cells. You can also modify the code to use
# a different tag word
c.TagRemovePreprocessor.remove_cell_tags = ("remove",)
c.TagRemovePreprocessor.remove_all_outputs_tags = ("remove_output",)
c.TagRemovePreprocessor.remove_input_tags = ("hide",)
c.TagRemovePreprocessor.enabled = True

# ensure the graphics are included in the html
c.HTMLExporter.exclude_output_prompt = True
c.HTMLExporter.exclude_input_prompt = True
c.HTMLExporter.embed_images = True

# Configure and run out exporter
c.HTMLExporter.preprocessors = ["nbconvert.preprocessors.TagRemovePreprocessor"]
exporter = HTMLExporter(config=c)
exporter.register_preprocessor(TagRemovePreprocessor(config=c), True)


# Configure and run our exporter - returns a tuple - first element with html,
# second with notebook metadata
output = HTMLExporter(config=c).from_filename(notebook_file_name + ".ipynb")

# Write to output html file
with open(notebook_file_name + ".html", "w") as f:
    f.write(output[0])

# open the file with the operating system
import os

errorCode = os.system("open " + notebook_file_name + ".html")
```
