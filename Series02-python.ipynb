# python-series02 : Sinusoids, from math to code 

To generate a sinusoidal signal, the math reads : 


$y(t) = A \cdot \sin(2 \pi f t + \phi)$

where :
- $A$ is the amplitude (peak value)
- $f$ is the frequency in Hertz (cycles per second)
- $t$ is the time in seconds
- $φ$ is the phase in radians (0 to 2π)
- $s(t)$ is the signal value at time t

```python
import numpy as np
import matplotlib.pyplot as plt

# import the necessary libraries
# create a numpy array called t that contains values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)

# create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = 1 * np.sin(2 * np.pi * 1 * t + 0)  

# plot the signal s as a function of time t
plt.plot(t, s)
plt.xlabel('Time [s]')
plt.ylabel('amplitude (unit?s)')
plt.title('Sinusoidal function')
plt.grid()
plt.savefig('sine_wave.png')
plt.show()
```

```python
# create a cos wave 
c = 1 * np.cos(2 * np.pi * 1 * t + 0)
plt.plot(t, c, label = "cos" , color='purple')
plt.plot(t, s, label = "sin" , color='orange')
# Add a second plot of a cosine function of t in the same figure

# Add a legend to the plot
plt.legend(['sine', 'cosine'])


plt.grid()


# Add a second label to the y axis
plt.ylabel('amplitude (unit?s)')


# Save the plot as a pdf file
plt.savefig('cosine_wave.pdf')
plt.show()
```

```python
# Create a function called sinusoid that :
#.      takes as input all parameters that you need to define the sinusoid
#.      returns a numpy array called x
def sinusoid(amplitude, frequency, phase, time):
    x = amplitude * np.sin(2 * np.pi * frequency * time + phase)
    return x
# Use the function sinusoid to create two numpy arrays s1 and s2 with the two different frequencies
s1 = sinusoid(1, 1, 0, t)
s2 = sinusoid(1, 2, 0, t)

# Plot s1 and s2 as a function of t (as illustrated in the graph below
plt.plot(t, s1, label = "sinusoid 1" , color='purple')
plt.plot(t, s2, label = "sinusoid 2" , color='orange')

plt.legend()
plt.grid()
plt.xlabel('Time [s]')
plt.ylabel('amplitude (unit?s)')

# Save the plot as a pdf file
# plt.savefig('s1_s2_wave.pdf')
plt.show()
```
