# python-series02 : Sinusoids, from math to code 

To generate a sinusoidal ssignal, the math reads : 


$y(t) = A \cdot \sin(2 \pi f t + \phi)$

where :
- $`A`$ is the amplitude (peak value)
- $`f`$ is the frequency in Hertz (cycles per second)
- $`t`$ is the time in seconds
- $`φ`$ is the phase in radians (0 to 2π)
- $`s(t)`$ is the signal value at time `t`

```python
import numpy as np
import matplotlib.pyplot as plt

# import the necessary libraries
# create a numpy array called t that contains values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)

# create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = 1 * np.sin(2 * np.pi * 1 * t + 0)  

# plot the signal s as a function of time t
plt.plot(t, s)
plt.xlabel('Time [s]')
plt.ylabel('amplitude (unit?s)')
plt.title('Sinusoidal function')
plt.grid()
plt.savefig('sine_wave.png')
plt.show()
```
